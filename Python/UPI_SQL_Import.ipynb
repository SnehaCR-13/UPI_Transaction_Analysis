### Objective
The objective of this notebook is to import the cleaned UPI datasets into a MySQL database and verify successful data ingestion. This notebook demonstrates the ETL (Extract, Transform, Load) process by reading CSV files using Python, performing necessary preprocessing, and loading the data into SQL tables for further analysis.

### Tools Used
- Python
- Pandas
- MySQL
- SQLAlchemy
- PyMySQL

### Workflow
1. Import required libraries
2. Read CSV datasets
3. Perform basic data cleaning and preprocessing
4. Connect Python to MySQL
5. Create SQL tables (already created using DDL scripts in MySQL)
6. Import datasets into SQL
7. Validate imported records
8. Verify row counts and data integrity
